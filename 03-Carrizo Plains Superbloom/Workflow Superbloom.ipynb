## Project Workflow
In this project, I focused on creating a map of iNaturalist entries in the Carrizo Plains area. While I wrote a bit of Python to download and organize the data I wanted (plants in the area), this project did not include extensive geoprocessing. Instead I focused on the layout feature of ArcGIS, learning all the ins and outs of text and shape formatting within pro's UI.

After some exploration into iNaturalist's API protocols, I decided to pull only research-grade entries from Feb-Apr 2026 as a way to keep the map symbology a little bit cleaner. I used a rectangular bounding box for the data pull (35.0 N, 119.9 W to 35.4 N, 119.5 W), but ended up clipping that down to a more organic Carrizo Plains shape using a vegetation layer from the CA Open Data Portal here: https://data.ca.gov/dataset/vegetation-carrizo-plain-national-monument-ecological-reserve-and-adjacent-elk-range-ds10941. I also dissolved this layer to create a Carrizo Plains boundary that I used on my final map.

While thinking of elements to add to a map, I decided to use pictures from iNaturalist on a legend for each plant type. Obviously there are lot of plants, so I wrote a script to copy a layer with only the top 12 plants by observation count. I changed symbol shapes and colors to differentiate between sightings, ordered them by rank so that the least seen plants appear at the top of the layer order, and created a layout with an appropriate extent and center.

Originally I wanted my layout legend to include my pictures, but after messing around with it I realized they're just too tall relative to a line of text. Pivoted to numbering, with a second legend a grid of numbered pictures. After some very precise (and tedious) sizing and positioning, I had something that looks pretty good.
