### Project and Discussion
In this project I continued my work with raster dataset processing to create a suitability layer for new solar development. This suitability analysis included known technoeconomic and protected area exclusion zones, topography, and solar irradiance. In previous projects I included a more robust narrative, but I feel that's becoming less necessary as I become more competent in ArcGIS.

The geoprocessing in this project was pretty straightforward, leaving data gathering and organization the main challenge. Like in previous projects, I started by downloading DEM links from USGS: https://apps.nationalmap.gov/downloader. I scrubbed for most recent data in my area of interest, ran a script to download the data, and used the mosaictonewraster method to generate an elevation raster. I then cleaned up the raster and reclassified it based on slope; 5 being the flattest, 1 being the least flat (suitability_slope).

Next I took an exclusions layer from the CA Open Data Portal: https://data.ca.gov/dataset/base-exclusions-solar. I converted the polygon feature class into a raster and reclassified it so exclusions = 0, no exclusion (or no data) = 1. I did the binary this way so I could just multiply this raster and the suitability rasters together to get a flat 0 if there are exclusions in the area.

Finally I found irradiance data for my study area in the National Solar Radiation Database: https://nsrdb.nlr.gov. I downloaded csvs for the area and used pandas dataframe methods to create a new csv, appending location and average irradiance data for all the downloaded data. I created a feature class from the new csv and converted these points into a raster using the Inverse Distance Weighted method, then reclassified the raster based on value; 5 being the highest GHI, 1 being the least GHI (suitability_irradiance).

Combining the data was easy enough; I decided to use Pro's UI here instead of Python. I multiplied the exclusions raster and the slope raster to get an intermediate suitability_combined, then used a weighted sum on that and the irradiance raster weighting equally 0.5 and 0.5. This got me suitability_final, a raster rating solar suitability from 1 (worst irradiance and slope) to 5 (best irradiance and slope), excluding any exclusion zones.

Finally, I clipped everything I had down to the three counties in the area and put some borders on there to show the county lines and the area my data covered. It doesn't over all the way to Arizona, as I was really interested in land close to San Diego. There area few discrepancies in the data; a few cells in the Salton Sea as well as the San Diego Bay weren't excluded in the exclusion data, but obviously these aren't suitable for solar development.

Results are pretty typical; the main problem with this project is that despite using solar irradiance data as a parameter for suitability, I suspect there is some consideration missing for peaks and valleys in the elevation. For example, Barona Valley has a solid stretch of cells with a suitability of 3+, but I have a hard time believing that this area gets better irradiance than, say, the San Diego Bay (which scores between 2 and 3 on suitability, allegedly). I would love to follow up on this if solar suitability was my job, but alas.
